```python
import getpass
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
model = ChatOpenAI(model_name = "gpt-3.5-turbo")
```

```python
from langgraph.graph.message import add_messages
from langgraph.graph import StateGraph, START, END
```

```python
from typing import TypedDict, Annotated, List
from typing_extensions import TypedDict  
class State(TypedDict):
    messages: Annotated[List, add_messages]
```

```python
def bot(state: State):
    print(f"In bot, using State class: {state['messages']}")
    return {'messages': [model.invoke(state['messages'])]}
```

```python
graph_builder = StateGraph(State)

graph_builder.add_node("bot", bot)

graph_builder.set_entry_point('bot')
graph_builder.set_finish_point('bot')

graph = graph_builder.compile()
```

```python
res = graph.invoke({'messages':['Hello, how are you?']})
print(res['messages'])
```

```python
while True:
    user_ip = input("User :")
    if user_ip.lower() in['quit','bye','exit','q','e']:
        print("Assistant: GoodBye, Thank You :)")
        break
    for event in graph.stream({'messages':("user",user_ip)}):
        for value in event.values():
            print("Assistant: ", value['messages'][-1].content)
```
